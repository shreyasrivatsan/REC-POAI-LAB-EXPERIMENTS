```python
#1. Analyse the Exploration and Exploitation Trade-offs
import random

# Fitness function
def fitness(x):
    return max(
        -(x - 2) ** 2 + 10,
        -(x - 7) ** 2 + 15
    )


# Generate a random state
def explore():
    return random.randint(0, 10)


# Look for a better nearby state
def exploit(x):
    neighbors = []

    if x > 0:
        neighbors.append(x - 1)

    if x < 10:
        neighbors.append(x + 1)

    best = x

    for n in neighbors:
        if fitness(n) > fitness(best):
            best = n

    return best


# Start from a random state
current = explore()

print("Initial State:", current)
print("Initial Fitness:", fitness(current))

for i in range(10):

    # Exploitation
    new_state = exploit(current)

    if fitness(new_state) > fitness(current):
        current = new_state

    # Exploration
    if random.random() < 0.3:
        current = explore()

    print("Iteration", i + 1,
          "State:", current,
          "Fitness:", fitness(current))


print("\nFinal State:", current)
print("Final Fitness:", fitness(current))
```

```python
#2. Hill Climbing Search for Eight Queens Problem
import random

# Number of queens
N = 8


# Count the number of attacking pairs
def conflicts(board):
    count = 0

    for i in range(N):
        for j in range(i + 1, N):

            # Same row
            if board[i] == board[j]:
                count += 1

            # Same diagonal
            elif abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count


# Generate a random board
def create_board():
    board = []

    for i in range(N):
        board.append(random.randint(0, N - 1))

    return board


# Find the best neighboring state
def get_best_neighbor(board):
    best_board = board
    best_conflicts = conflicts(board)

    for col in range(N):

        for row in range(N):

            if row == board[col]:
                continue

            new_board = board.copy()
            new_board[col] = row

            new_conflicts = conflicts(new_board)

            if new_conflicts < best_conflicts:
                best_board = new_board
                best_conflicts = new_conflicts

    return best_board, best_conflicts


# Hill Climbing
board = create_board()

print("Initial Board:", board)
print("Initial Conflicts:", conflicts(board))

while True:

    current_conflicts = conflicts(board)

    # Solution found
    if current_conflicts == 0:
        break

    # Find a better neighboring state
    new_board, new_conflicts = get_best_neighbor(board)

    # No improvement possible
    if new_conflicts >= current_conflicts:
        print("Reached a local optimum!")
        break

    board = new_board

    print("Board:", board)
    print("Conflicts:", new_conflicts)


print("\nFinal Board:", board)
print("Final Conflicts:", conflicts(board))
```

```python
#3. Simulated Annealing Search
#Team 22: VLSI Layout Problem
import random
import math

# Components in the VLSI circuit
components = ["A", "B", "C", "D"]

# Connections between components
connections = [
    ("A", "B"),
    ("A", "C"),
    ("B", "D"),
    ("C", "D")
]

# Possible positions on the chip
positions = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]


# Calculate total wire length
def wire_length(layout):
    total = 0

    for a, b in connections:
        x1, y1 = layout[a]
        x2, y2 = layout[b]

        # Manhattan distance
        total += abs(x1 - x2) + abs(y1 - y2)

    return total


# Create a random layout
def create_layout():
    random_positions = positions.copy()
    random.shuffle(random_positions)

    layout = {}

    for i in range(len(components)):
        layout[components[i]] = random_positions[i]

    return layout


# Create a neighboring layout
def get_neighbor(layout):
    new_layout = layout.copy()

    a, b = random.sample(components, 2)

    # Swap two components
    new_layout[a], new_layout[b] = (
        new_layout[b],
        new_layout[a]
    )

    return new_layout


# Simulated Annealing
current_layout = create_layout()
current_cost = wire_length(current_layout)

best_layout = current_layout.copy()
best_cost = current_cost

temperature = 100

while temperature > 0.1:

    # Create a neighboring layout
    new_layout = get_neighbor(current_layout)

    new_cost = wire_length(new_layout)

    # Difference in cost
    difference = new_cost - current_cost

    # Accept a better layout
    if difference < 0:
        current_layout = new_layout
        current_cost = new_cost

    # Sometimes accept a worse layout
    else:
        probability = math.exp(-difference / temperature)

        if random.random() < probability:
            current_layout = new_layout
            current_cost = new_cost

    # Store the best solution found
    if current_cost < best_cost:
        best_layout = current_layout.copy()
        best_cost = current_cost

    # Reduce temperature
    temperature *= 0.95


print("Best VLSI Layout:")
print(best_layout)

print("Minimum Wire Length:", best_cost)
```
