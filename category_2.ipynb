```python
# Fully Observable Environment

rooms = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Dirty",
    "D": "Clean"
}

print("Initial Room Status:")
print(rooms)

for room in rooms:
    if rooms[room] == "Dirty":
        print(f"Cleaning Room {room}")
        rooms[room] = "Clean"
    else:
        print(f"Room {room} is already clean.")

print("\nFinal Room Status:")
print(rooms)
```

```python
# Partially Observable Environment

rooms = {
    "A": "Dirty",
    "B": "Clean",
    "C": "Dirty",
    "D": "Clean"
}

memory = {}

visit_order = ["A", "B", "C", "D"]

print("Robot starts cleaning...\n")

for room in visit_order:
    status = rooms[room]

    print(f"Robot enters Room {room}")

    memory[room] = status

    if status == "Dirty":
        print(f"Cleaning Room {room}")
        rooms[room] = "Clean"
        memory[room] = "Clean"
    else:
        print(f"Room {room} is already clean.")

    print("Memory:", memory)
    print()

print("Final Room Status:")
print(rooms)
```

```python
# Known Environment

graph = {
    'A': ['B'],
    'B': ['A', 'C'],
    'C': ['B', 'D'],
    'D': ['C']
}

start = 'A'
goal = 'D'

path = ['A', 'B', 'C', 'D']

print("Known Environment")
print("Start:", start)
print("Goal:", goal)
print("Path:", " -> ".join(path))
print("Goal Reached!")
```

```python
# Unknown Environment

graph = {
    'A': ['B'],
    'B': ['A', 'C'],
    'C': ['B', 'D'],
    'D': []
}

start = 'A'
goal = 'D'

visited = []
queue = [start]

print("Unknown Environment")

while queue:
    node = queue.pop(0)

    if node not in visited:
        print("Exploring:", node)
        visited.append(node)

        if node == goal:
            print("Goal Found!")
            break

        queue.extend(graph[node])
```

```python
# Static Environment

position = 0
goal = 5
obstacle = 3

print("Static Environment")

while position < goal:
    position += 1

    if position == obstacle:
        print(f"Obstacle at {position}. Going around it.")
        continue

    print(f"Moved to position {position}")

print("Goal Reached!")
```

```python
# Dynamic Environment

position = 0
goal = 5
obstacle = 3

print("Dynamic Environment")

while position < goal:
    obstacle += 1      # obstacle keeps moving

    position += 1

    if position == obstacle:
        print(f"Obstacle moved to {obstacle}. Waiting...")
        continue

    print(f"Moved to position {position}")

print("Goal Reached!")
```

```python
# Discrete Environment

position = 0
goal = 5

print("Discrete Environment")

while position < goal:
    position += 1
    print(f"Car moved to position {position}")

print("Destination Reached!")
```

```python
# Continuous Environment

position = 0.0
goal = 5.0
speed = 0.5

print("Continuous Environment")

while position < goal:
    position += speed
    print(f"Car position: {position:.1f}")

print("Destination Reached!")
```

```python
# Deterministic Environment

destination = "Home"

print("Deterministic Environment")

for step in range(1, 6):
    print(f"Step {step}: Moving towards {destination}")

print("Package Delivered Successfully!")
```

```python
# Stochastic Environment

import random

destination = "Home"

print("Stochastic Environment")

for step in range(1, 6):
    if random.choice([True, False]):
        print(f"Step {step}: Obstacle encountered! Finding another path.")
    else:
        print(f"Step {step}: Moving towards {destination}")

print("Package Delivered!")
```

```python
# Episodic Environment

fruits = ["Good", "Bad", "Good", "Bad", "Good"]

print("Episodic Environment")

for i, fruit in enumerate(fruits, start=1):
    if fruit == "Good":
        print(f"Fruit {i}: Picked")
    else:
        print(f"Fruit {i}: Rejected")
```

```python
# Sequential Environment

position = 0
goal = 5

print("Sequential Environment")

while position < goal:
    position += 1
    print(f"Moved to position {position}")

print("Goal Reached!")
```
