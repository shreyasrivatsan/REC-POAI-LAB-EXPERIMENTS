```python
# Bayesian Reasoning - Medical Diagnosis

prior_disease = 0.01
sensitivity = 0.90
specificity = 0.95

positive_test = sensitivity * prior_disease
false_positive = (1 - specificity) * (1 - prior_disease)

probability = positive_test / (positive_test + false_positive)

print("Medical Diagnosis using Bayesian Reasoning")
print("Probability of disease:", round(probability * 100, 2), "%")

if probability > 0.5:
    print("High probability of disease")
else:
    print("Low probability of disease")
```

```python
# Fuzzy Inference - Medical Diagnosis

fever = float(input("Enter fever level (0-100): "))
cough = float(input("Enter cough severity (0-100): "))

low_fever = max(0, min(1, (50 - fever) / 50))
high_fever = max(0, min(1, (fever - 30) / 40))

mild_cough = max(0, min(1, (50 - cough) / 50))
severe_cough = max(0, min(1, (cough - 30) / 40))

low_risk = min(low_fever, mild_cough)
high_risk = max(high_fever, severe_cough)

print("Medical Diagnosis using Fuzzy Logic")

if high_risk > low_risk:
    print("High health risk")
else:
    print("Low health risk")
```

```python
# Bayesian Reasoning - Autonomous Vehicle Obstacle Detection

prior_obstacle = 0.20

radar_detection = 0.90
camera_detection = 0.85

radar_no_obstacle = 0.10
camera_no_obstacle = 0.15

probability = (prior_obstacle * radar_detection * camera_detection) / (
    (prior_obstacle * radar_detection * camera_detection) +
    ((1 - prior_obstacle) * radar_no_obstacle * camera_no_obstacle)
)

print("Autonomous Vehicle Obstacle Detection")
print("Probability of obstacle:", round(probability * 100, 2), "%")

if probability > 0.5:
    print("Obstacle detected! Apply braking.")
else:
    print("No immediate obstacle detected.")
```

```python
# Fuzzy Inference - Autonomous Vehicle Braking System

distance = float(input("Enter distance from obstacle (meters): "))
speed = float(input("Enter vehicle speed (km/h): "))

near = max(0, min(1, (30 - distance) / 25))
far = max(0, min(1, (distance - 10) / 30))

high_speed = max(0, min(1, (speed - 30) / 70))
low_speed = max(0, min(1, (60 - speed) / 60))

emergency = min(near, high_speed)
normal = max(far, low_speed)

print("Autonomous Vehicle Fuzzy Braking System")

if emergency > normal:
    print("Emergency braking required")
else:
    print("Normal braking or safe driving")
```

```python
# Bayesian Reasoning - Smart Manufacturing Machine Fault Detection

prior_fault = 0.10

vibration_high = 0.80
temperature_high = 0.90

vibration_normal = 0.20
temperature_normal = 0.10

probability = (prior_fault * vibration_high * temperature_high) / (
    (prior_fault * vibration_high * temperature_high) +
    ((1 - prior_fault) * vibration_normal * temperature_normal)
)

print("Smart Manufacturing - Machine Fault Detection")
print("Probability of machine fault:", round(probability * 100, 2), "%")

if probability > 0.5:
    print("Machine fault detected. Maintenance required.")
else:
    print("Machine operating normally.")
```

```python
# Fuzzy Inference - Smart Manufacturing Predictive Maintenance

vibration = float(input("Enter vibration level (0-100): "))
temperature = float(input("Enter temperature level (0-100): "))

low_vibration = max(0, min(1, (50 - vibration) / 50))
high_vibration = max(0, min(1, (vibration - 30) / 40))

low_temperature = max(0, min(1, (50 - temperature) / 50))
high_temperature = max(0, min(1, (temperature - 30) / 40))

normal = min(low_vibration, low_temperature)
maintenance = max(high_vibration, high_temperature)

print("Smart Manufacturing using Fuzzy Logic")

if maintenance > normal:
    print("Maintenance required")
else:
    print("Machine operating normally")
```
