```python
def minimax(depth, node, maximizing, scores):
    if depth == 2:
        return scores[node]

    if maximizing:
        best = float('-inf')
        for i in range(2):
            best = max(best, minimax(depth + 1,
                                     node * 2 + i,
                                     False, scores))
        return best
    else:
        best = float('inf')
        for i in range(2):
            best = min(best, minimax(depth + 1,
                                     node * 2 + i,
                                     True, scores))
        return best

scores = [3, 5, 2, 9]
print("Best score:", minimax(0, 0, True, scores))
```

```python
def alphabeta(depth, node, maximizing, scores, alpha, beta):
    if depth == 2:
        return scores[node]

    if maximizing:
        best = float('-inf')

        for i in range(2):
            value = alphabeta(depth + 1, node * 2 + i,
                              False, scores, alpha, beta)
            best = max(best, value)
            alpha = max(alpha, best)

            if alpha >= beta:
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            value = alphabeta(depth + 1, node * 2 + i,
                              True, scores, alpha, beta)
            best = min(best, value)
            beta = min(beta, best)

            if alpha >= beta:
                break

        return best

scores = [3, 5, 2, 9]

print("Best score:", alphabeta(
    0, 0, True, scores, float('-inf'), float('inf')
))
```

```python
def print_board(board):
    for i in range(0, 9, 3):
        print(board[i], "|", board[i+1], "|", board[i+2])
    print()

def winner(board, player):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] == player:
            return True
    return False

def minimax(board, maximizing):
    if winner(board, 'O'):
        return 1
    if winner(board, 'X'):
        return -1
    if ' ' not in board:
        return 0

    if maximizing:
        best = -float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                best = max(best, minimax(board, False))
                board[i] = ' '
        return best
    else:
        best = float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                best = min(best, minimax(board, True))
                board[i] = ' '
        return best

board = [' '] * 9

while True:
    print_board(board)

    move = int(input("Enter your position (1-9): ")) - 1

    if move < 0 or move > 8 or board[move] != ' ':
        print("Invalid move! Try again.")
        continue

    board[move] = 'X'

    if winner(board, 'X'):
        print_board(board)
        print("You win!")
        break

    if ' ' not in board:
        print_board(board)
        print("Draw!")
        break

    best_score = -float('inf')
    best_move = -1

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            score = minimax(board, False)
            board[i] = ' '

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = 'O'
    print("Computer chooses position:", best_move + 1)

    if winner(board, 'O'):
        print_board(board)
        print("Computer wins!")
        break

    if ' ' not in board:
        print_board(board)
        print("Draw!")
        break
```
