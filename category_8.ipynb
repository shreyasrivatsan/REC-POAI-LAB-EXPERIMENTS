```python
facts = {"fever", "cough", "body_pain"}

rules = {
    "flu": ["fever", "cough", "body_pain"],
    "cold": ["cough", "sneezing"],
    "infection": ["fever", "body_pain"]
}

def forward_chaining(facts):
    facts = facts.copy()

    while True:
        new_fact = False

        for result, conditions in rules.items():
            if all(c in facts for c in conditions):
                if result not in facts:
                    facts.add(result)
                    print("Derived:", result)
                    new_fact = True

        if not new_fact:
            break

    return facts

def backward_chaining(goal, facts):
    if goal in facts:
        return True

    if goal in rules:
        return all(backward_chaining(c, facts)
                   for c in rules[goal])

    return False

print("Forward Chaining:")
final_facts = forward_chaining(facts)

print("\nBackward Chaining:")
print("Flu:", backward_chaining("flu", facts))
```

```python
facts = {"red_light", "pedestrian"}

rules = {
    "stop": ["red_light"],
    "slow_down": ["pedestrian"],
    "move": ["green_light", "no_obstacle"]
}

def forward_chaining(facts):
    conclusions = set(facts)

    for result, conditions in rules.items():
        if all(c in conclusions for c in conditions):
            conclusions.add(result)
            print("Action:", result)

    return conclusions

def backward_chaining(goal, facts):
    if goal in facts:
        return True

    if goal in rules:
        return all(backward_chaining(c, facts)
                   for c in rules[goal])

    return False

print("Forward Chaining:")
final_facts = forward_chaining(facts)

print("\nBackward Chaining:")
print("Should stop?", backward_chaining("stop", facts))
print("Can move?", backward_chaining("move", facts))
```
